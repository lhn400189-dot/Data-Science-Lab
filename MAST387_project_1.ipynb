### Problem A-3

```python
def Canadish_12_case(sentence):
    vowels = ['a', 'e', 'i', 'o', 'u', 'y']
    words = sentence.split()
    modified_words = []

    for word in words:
        # Check capitalization
        is_capitalized = word[0].isupper()
        is_all_caps = word.isupper()

        # work on lowercase
        lw = word.lower()

        # apply rule 1 or 2
        if lw[0] in vowels:
            mw = lw + "eh"
        else:
            index = 0
            while index < len(lw) and lw[index] not in vowels:
                index += 1
            if index < len(lw):
                mw = lw[index:] + lw[:index] + "ey"
            else:
                mw = lw + "ey"

        # restore capitalization
        if is_all_caps:
            mw = mw.upper()
        elif is_capitalized:
            mw = mw.capitalize()

        modified_words.append(mw)

    # Join words
    return " ".join(modified_words)

# test
input_sentence = "The weather in Montreal is AWESOME"
output_sentence = Canadish_12_case(input_sentence)
print(output_sentence)
```

### Problem A-4

```python
def Canadish(sentence):
    vowels = ['a', 'e', 'i', 'o', 'u', 'y']

    def translate_word(word):
        # same translation logic as A-3
        is_capitalized = word[0].isupper()
        is_all_caps = word.isupper()
        lw = word.lower()

        if lw[0] in vowels:
            mw = lw + "eh"
        else:
            index = 0
            while index < len(lw) and lw[index] not in vowels:
                index += 1
            if index < len(lw):
                mw = lw[index:] + lw[:index] + "ey"
            else:
                mw = lw + "ey"

        if is_all_caps:
            mw = mw.upper()
        elif is_capitalized:
            mw = mw.capitalize()

        return mw

    # Process sentence character by character
    result = []
    current_word = ""

    for ch in sentence:
        if ch.isalpha():
            current_word += ch
        else:
            # finish a word if we just ended one
            if current_word:
                result.append(translate_word(current_word))
                current_word = ""
            result.append(ch)

    # flush last word if sentence ends with letter
    if current_word:
        result.append(translate_word(current_word))

    return "".join(result)

# Test
input_sentence = "Here is a Canadish sentence with bells and whistles--YES, this is it! You made it."
output_sentence = Canadish(input_sentence)
print(output_sentence)
```

### Problem B-5

```python
import numpy as np
import seaborn.objects as so

def test_matrix(n, k, c):
    A = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            if i == j:
                A[i, j] = k ** c
            elif abs(i - j) <= k:
                A[i, j] = k - abs(i - j)
    return A


def steepest_GD(A, b, TOL, Nmax):
    n = len(b)
    x = np.zeros(n)
    r = b - np.dot(A, x)
    residuals = [np.linalg.norm(r)]
    i = 0
    while np.linalg.norm(r) > TOL and i < Nmax:
        Ar = np.dot(A, r)
        alpha = np.dot(r, r) / np.dot(r, Ar)
        x += alpha * r
        r = b - np.dot(A, x)
        residuals.append(np.linalg.norm(r))
        i += 1
    return residuals


# Parameters
n = 100
ks = [5, 10, 15]
cs = [1, 2, 3]
TOL = 1e-4
Nmax = 1000
b = np.ones(n)

# Plotting
records = []
for k in ks:
    for c in cs:
        A = test_matrix(n, k, c)
        residuals = steepest_GD(A, b, TOL, Nmax)
        for i, r in enumerate(residuals):
            records.append({
                "iteration": i,
                "residual": r,
                "label": f"k={k}, c={c}"
            })

df = pd.DataFrame(records)

(
    so.Plot(df, x="iteration", y="residual", color="label")
    .add(so.Line())
    .scale(x="log", y="log")
    .label(
        title="Convergence of Residuals for Various $k$ and $c$",
        x="Iteration ($i$)",
        y="Norm of Residual ($\\|r_i\\|$)",
        color="(k, c)"
    )
    .theme({
        "figure.figsize": (12, 8),
        "axes.titlesize": 16,
        "axes.labelsize": 14,
        "legend.fontsize": 11,
        "legend.loc": "upper right",
        "lines.linewidth": 2,
        "xtick.labelsize": 11,
        "ytick.labelsize": 11
    })
    .show()
)
```

The residual norm converges fastest when k = 5 and c = 3, meaning a smaller bandwidth and larger diagonal weight make convergence quicker.
The slowest convergence occurs when k = 15 and c = 1.
No combination reached the maximum number of iterations (Nmax = 1000); all cases converged to the tolerance TOL = 1e-4 before the limit.

### Problem B-6

```python
import numpy as np
import seaborn.objects as so

def test_matrix(n, k, c):
    A = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            if i == j:
                A[i, j] = k ** c
            elif abs(i - j) <= k:
                A[i, j] = k - abs(i - j)
    return A


def steepest_GD(A, b, TOL, Nmax):
    n = len(b)
    x = np.zeros(n)
    r = b - np.dot(A, x)
    residuals = [np.linalg.norm(r)]
    i = 0

    while np.linalg.norm(r) > TOL and i < Nmax:
        Ar = np.dot(A, r)
        alpha = np.dot(r, r) / np.dot(r, Ar)
        x += alpha * r
        r = b - np.dot(A, x)
        residuals.append(np.linalg.norm(r))
        i += 1

    return residuals


# Parameters
n = 100
ks = [5, 10, 15]
cs = [1, 2, 3]
TOL = 1e-4
Nmax = 1000
b = np.ones(n)

# Compute & plot
condition_numbers = {}

# Collect data into a DataFrame
records = []
for k in ks:
    for c in cs:
        A = test_matrix(n, k, c)
        cond_number = np.linalg.cond(A)
        condition_numbers[(k, c)] = cond_number
        residuals = steepest_GD(A, b, TOL, Nmax)
        label = f"k={k}, c={c}, cond={cond_number:.2f}"
        for i, r in enumerate(residuals):
            records.append({"iteration": i, "residual": r, "label": label})

df = pd.DataFrame(records)

# Plot style
(
    so.Plot(df, x="iteration", y="residual", color="label")
    .add(so.Line())
    .scale(x="log", y="log")
    .label(
        title="Convergence of Residuals with Condition Numbers",
        x="Iteration ($i$)",
        y="Norm of Residual ($\\|r_i\\|$)",
        color="(k, c, cond)"
    )
    .theme({
        "figure.figsize": (12, 8),
        "axes.titlesize": 16,
        "axes.labelsize": 14,
        "legend.fontsize": 10,
        "legend.loc": "upper right",
        "lines.linewidth": 2,
        "xtick.labelsize": 11,
        "ytick.labelsize": 11
    })
    .show()
)

# Print condition numbers
for (k, c), cond_number in condition_numbers.items():
    print(f"Condition number for k={k}, c={c}: {cond_number:.2f}")
```

Yes, a smaller condition number corresponds to faster convergence.
From the plot, cases with low condition numbers (e.g., k=5, c=3 or k=15, c=3) converge rapidly,
while matrices with large condition numbers (e.g., k=15, c=1) converge much slower.
This confirms that lower condition numbers lead to better numerical stability and faster convergence.
